```python
import duckdb
from faker import Faker
import pandas as pd

# Config
db_file = "test.duckdb"

def get_df(count):
    # Create fake data
    fake = Faker()
    records = []
    for cid in range(1, count + 1):
        first_name = fake.first_name()
        last_name = fake.last_name()
        email = fake.company_email()
        records.append({
            "customer_id": cid,
            "first_name": first_name,
            "last_name": last_name,
            "email": email
        })

    # Put into DataFrame
    df = pd.DataFrame(records)
    return df

# Save to DuckDB
con = duckdb.connect(db_file)
df = get_df(1000)
con.execute("CREATE TABLE customers_sm AS SELECT * FROM df")
# df = get_df(10000)
# con.execute("CREATE TABLE customers_md AS SELECT * FROM df")
# df = get_df(100000)
# con.execute("CREATE TABLE customers_lg AS SELECT * FROM df")
con.close()
```
