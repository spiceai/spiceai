```python
import boto3

# Use a profile by name
session = boto3.Session(profile_name='spice-sandbox', region_name="ap-southeast-2")
dynamodb = session.resource('dynamodb')

# Create 5000 random rows in a table
table = dynamodb.Table("orders")

# Table schema:
# sql> describe orders_stream;
# +-------------+-----------+-------------+
# | column_name | data_type | is_nullable |
# +-------------+-----------+-------------+
# | amount      | Float64   | YES         |
# | customer    | Utf8      | YES         |
# | id          | Utf8      | YES         |
# | status      | Utf8      | YES         |
# +-------------+-----------+-------------+
import random
import string
items = []
for i in range(5000):
    # Float types are not supported. Use Decimal types instead.
    item = {
        'id': ''.join(random.choices(string.ascii_lowercase + string.digits, k=10)),
        'customer': 'customer_' + str(random.randint(1, 100)),
        'amount': round(random.uniform(10.0, 1000.0), 2),
        'status': random.choice(['pending', 'shipped', 'delivered', 'canceled'])
    }

    # Update amount to be Decimal
    from decimal import Decimal
    item['amount'] = Decimal(str(item['amount']))

    items.append(item)


with table.batch_writer() as batch:
    for item in items:
        batch.put_item(Item=item)
```
